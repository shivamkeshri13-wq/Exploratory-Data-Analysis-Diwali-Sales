# **EXploratory Data Analysis**

# *Gender*

```python
# plotting a bar chart for Gender and it's count

ax = sns.countplot(x = 'Gender',data = df)
plt.title('Total Orders by Gender')

for bars in ax.containers:
      ax.bar_label(bars)
      plt.show()
```

```python
# plotting a bar chart for gender vs total amount
```

```python
sales_gen = df.groupby(['Gender'], as_index = False)['Amount'].sum().sort_values(by='Amount', ascending=False)
bx = sns.barplot(x = 'Gender',y= 'Amount' ,data = sales_gen)
    
plt.title('Total Sales by Gender')
plt.xlabel('Gender')
plt.ylabel('Total Amount')
plt.show()
```

**From above graphs we can see that most of the buyers are females and even the purchasing power of females are greater than men**

```python
bx = sns.countplot(data = df,x ='Age Group',hue = 'Gender' )

for containers in bx.containers:
     bx.bar_label(containers)

plt.title('Total Order By Age Group')
plt.show()
```

```python
sale_age = df.groupby(['Age Group'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False)

sns.barplot(x = 'Age Group', y = 'Amount', data = sale_age)
plt.title('Total Sales Amount By Age Group')
plt.show()
```

**From above graphs we can see that most of the buyers are of age group between 26-35 yrs female**

# *Total Number of orders from top 10 states*




```python
sales_states = df.groupby(['State'], as_index = False)['Orders'].sum().sort_values(by = 'Orders', ascending = False).head(10)
sns.set(rc={'figure.figsize':(22,5)})

sns.barplot(x = 'State', y = 'Orders', data = sales_states)
plt.title('Total Order By State')
plt.show()
```

# *Total Amount/sales from top 10 states*

```python
sales_states = df.groupby(['State'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False).head(10)
sns.barplot(x = 'State',y = 'Amount' , data =sales_states)
sns.set(rc={'figure.figsize':(22,5)})
plt.title('Total Amount By States')
plt.show()
```

**From above graphs we can see that most of the orders & total sales/amount are from Uttar Pradesh, Maharashtra and Karnataka respectively**
